#### Problem 4

```python
import pandas as pd
import numpy as np
from scipy.stats import norm
```

```python
R = 10000
seed = 42
```

```python
ames = pd.read_csv("/Users/rayyansiddiqui/Documents/DATA_SCIENCE/DS5030/UU_F26_templates/data/ames_prices.csv")
air = (ames["Central.Air"] == "Y").astype(int).to_numpy()
estimate = air.mean()
null = 0.90
rng = np.random.default_rng(42)
boot = []
for i in range(R):
    sample = rng.choice(air, len(air), replace=True)
    boot.append(sample.mean())
```

```python
se = np.std(boot, ddof=1)
z = (estimate - null) / se
p = 2 * norm.sf(abs(z))
print("Estimate:", round(estimate, 4), "SE:", round(se, 4), "z:", round(z, 4), "p:", p)
for level, c in [(90, 1.645), (95, 1.96), (99, 2.576)]:
    print(level, "% CI:", round(estimate-c*se, 4), round(estimate+c*se, 4))
```

D) Reject at all three confidence intervals.
E) All of the confidence intervals do not include .90 which is in agreement with D.
F) The proportion is 93% and we reject the null hypothesis due to the p value being extremely small. The proportion of homes with central air is significantly different from 90%.

### Problem 5

```python
pairs = np.arcsinh(ames[["area", "price"]].dropna().to_numpy())
estimate = np.corrcoef(pairs[:, 0], pairs[:, 1])[0, 1]
null = 0
rng = np.random.default_rng(42)
boot = []
for i in range(R):
    sample = pairs[rng.choice(len(pairs), len(pairs), replace=True)]
    boot.append(np.corrcoef(sample[:, 0], sample[:, 1])[0, 1])
```

```python
se = np.std(boot, ddof=1)
z = (estimate - null) / se
p = 2 * norm.sf(abs(z))
print("Estimate:", round(estimate, 4), "SE:", round(se, 4), "z:", round(z, 4), "p:", p)
for level, c in [(90, 1.645), (95, 1.96), (99, 2.576)]:
    print(level, "% CI:", round(estimate-c*se, 4), round(estimate+c*se, 4))
```

D) Reject null hypothesis for all three intervals
E) Each confidence interval do not include 0. 
F) The correlation between the two factors is .723. The CI for 95% is .7032, .7431. We reject null hypothesis because of the p value being incredibly small. Based off the question, the larger the home is, the higher the price.
